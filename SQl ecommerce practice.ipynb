```python
# 1. حول العمود في الباندا (هو سيعرف كيف يعالج 1/1 و 10/12 تلقائياً)
df = pd.read_sql_query("SELECT * FROM ecommerce", conn)
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# 2. الآن استخرجي الأقدم والأحدث بكل سهولة وبدقة 100%
print(df['InvoiceDate'].min())
print(df['InvoiceDate'].max())
```

```python
# الخطوة 1: سحب البيانات كما هي
query = "SELECT InvoiceDate FROM ecommerce"
df = pd.read_sql_query(query, conn)

# الخطوة 2: التحويل الاحترافي (هذا السطر هو سحر محلل البيانات)
# نستخدم errors='coerce' للتعامل مع أي بيانات خاطئة و dayfirst حسب صيغة ملفك
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')

# الخطوة 3: التحقق من "صحة" البيانات بعد التحويل
print(f"عدد القيم التي فشل تحويلها: {df['InvoiceDate'].isna().sum()}")
```

```python
# الخطوة 4: الحصول على النتائج بدقة 100%
oldest = df['InvoiceDate'].min()
latest = df['InvoiceDate'].max()

print(f"الأقدم: {oldest}")
print(f"الأحدث: {latest}")
```
